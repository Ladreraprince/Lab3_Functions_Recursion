```python
# [CELL 1] STUDENT CONFIGURATION
LAST_NAME = "Ladrera"
STUDENT_ID = "TUPM-26-1132"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Student:", LAST_NAME)
print("Seed Digit:", SEED_DIGIT)
print("Digit Sum:", ID_SUM)
print("Surname Length:", NAME_LENGTH)
```

```python
# [CELL 2] FUNCTIONAL ENCAPSULATION
def greet():
    print("=" * 40)
    print(f"System Initialized for: {LAST_NAME}")
    print("=" * 40)

def identity_code():
    code = SEED_DIGIT * NAME_LENGTH
    return code

greet()
print("Identity Code:", identity_code())
```

```python
# [CELL 3] LEGB VALIDATION
global_value = 100

def scope_test():
    global_value = SEED_DIGIT * 5  # Local shadowing
    print("Inside function (Local):", global_value)

scope_test()
print("Outside function (Global):", global_value)
```

```python
# [CELL 4] PARAMETRIC PROCESSING
def user_summary(title, *scores, **info):
    print(f"=== {title} ===")
    total = sum(scores)
    average = total / len(scores) if scores else 0
    print("Scores:", scores)
    print("Total:", total)
    print("Average:", round(average, 2))
    for k, v in info.items():
        print(f"{k}: {v}")
    return average

avg = user_summary(
    f"{LAST_NAME} Academic Report",
    SEED_DIGIT * 10,
    ID_SUM,
    NAME_LENGTH * 10,
    Id=STUDENT_ID,
    Surname=LAST_NAME
)
```

```python
# [CELL 5] RETURN MECHANISM
def compute_area(radius):
    area = 3.1416 * (radius ** 2)
    return area

calculated_area = compute_area(SEED_DIGIT)
print("Computed Area:", round(calculated_area, 2))
```

```python
# [CELL 6] SAFE CALCULATOR
def safe_calculator(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        result = "UNDEFINED"
    return result

print("Calculation Result:", safe_calculator(ID_SUM, SEED_DIGIT))
print("Zero Division Guard Test:", safe_calculator(ID_SUM, 0))
```

```python
# [CELL 7] RECURSION
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(f"Factorial of SEED_DIGIT ({SEED_DIGIT}):", factorial(SEED_DIGIT))
```

```python
# [CELL 8] STRING RECURSION
def reverse(text):
    if len(text) <= 1:
        return text
    return text[-1] + reverse(text[:-1])

print("Reversed Surname:", reverse(LAST_NAME.upper()))
```

```python
# [CELL 9] DECORATOR
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Executing: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def multiply(x, y):
    return x * y

print("Multiplication Output:", multiply(SEED_DIGIT, NAME_LENGTH))
```

```python
# [CELL 10] LAMBDA
data = list(range(1, SEED_DIGIT + 5))
squared = list(map(lambda x: x**2, data))

print("Initial Data Range:", data)
print("Squared Data:", squared)
```

```python
# [CELL 11] GENERATOR
def even_squares(limit):
    for i in range(limit):
        if i % 2 == 0:
            yield i**2

print("Generated Even Squares:")
for value in even_squares(NAME_LENGTH + SEED_DIGIT):
    print(value, end=" ")
print()
```
